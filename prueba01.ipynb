```python
%%shell
# Guarda el código C++ en un archivo temporal
cat > main.cpp << EOF
#include <iostream>
using namespace std;

int main()
{
    double a, b;

    cout <<"Hola Mundo" <<endl;
    cout << "Ingrese dos números: ";

    if (!(cin >> a >> b))
    {
        cout << "Entrada inválida. " <<endl;
    }
    else
    {
        double suma = a + b;
        cout << "La suma es: " << suma << endl;
    }
    return 0;
}
EOF

# Compila el código C++ usando g++
g++ main.cpp -o main

# Ejecuta el programa compilado
./main
```
